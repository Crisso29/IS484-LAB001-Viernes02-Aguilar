## Sección 5.2 — Ficha PEAS: Agente Termostato

| Elemento PEAS | Descripción |
|---------------|-------------|
| **P** — Desempeño | Mantener la temperatura lo más cercana posible a la temperatura objetivo |
| **E** — Entorno | Habitación con temperatura variable, sistema de aire acondicionado y calefacción |
| **A** — Actuadores | Encender aire acondicionado, encender calefacción, mantener sistema apagado |
| **S** — Sensores | Sensor de temperatura actual de la habitación |

## Sección 5.3 — Reglas de decisión

| Condición | Acción a tomar |
|-----------|----------------|
| Temperatura muy por encima de la deseada (diferencia > margen) | Encender aire acondicionado |
| Temperatura muy por debajo de la deseada (diferencia < -margen) | Encender calefacción |
| Temperatura cerca de la deseada | Mantener sistema apagado |

```python
## Sección 5.4 — Implementación del agente termostato

def agente_termostato(temperatura_actual, temperatura_objetivo=22, margen=1.0):
    """
    Agente reactivo simple sin memoria.
    P: Mantener temperatura cercana al objetivo.
    E: Habitación con aire acondicionado y calefacción.
    A: Encender AC, encender calefacción, mantener apagado.
    S: Sensor de temperatura actual.
    """
    diferencia = temperatura_actual - temperatura_objetivo

    if diferencia > margen:
        return "Encender aire acondicionado"
    elif diferencia < -margen:
        return "Encender calefacción"
    else:
        return "Mantener sistema apagado"
```

```python
## Sección 5.5 — Simulación

percepciones = [18.0, 19.5, 21.8, 24.3, 26.0, 23.0, 20.5, 21.0, 23.1]

for t in percepciones:
    accion = agente_termostato(t)
    print(f"Percepción: {t}°C -> Acción del agente: {accion}")
```

## Sección 5.6 — Reflexión

**¿El agente termostato es racional?**
Sí, es racional porque en cada momento elige la acción que maximiza 
su desempeño (mantener la temperatura objetivo) basándose únicamente 
en su percepción actual.

**¿Bajo qué condiciones dejaría de serlo?**
- Si el sensor falla y reporta temperaturas incorrectas, el agente 
  tomaría decisiones equivocadas (encender calefacción cuando ya hace calor).
- Si el margen es demasiado amplio, el agente toleraría temperaturas 
  muy alejadas del objetivo sin actuar, perdiendo su utilidad.
